```python
import pandas as pd
import numpy as np
from pathlib import Path

print("ShopSphere Data Cleaning & ETL")
print("=" * 50)
```

```python
RAW_PATH = Path("../02_Raw_Data")
CLEAN_PATH = Path("../04_Python_ETL/cleaned_data")

CLEAN_PATH.mkdir(parents=True, exist_ok=True)

print("Raw data:", RAW_PATH.resolve())
print("Clean data:", CLEAN_PATH.resolve())
```

```python
customers = pd.read_csv(RAW_PATH / "customers_raw.csv")
products = pd.read_csv(RAW_PATH / "products_raw.csv")
orders = pd.read_csv(RAW_PATH / "orders_raw.csv")
order_items = pd.read_csv(RAW_PATH / "order_items_raw.csv")
payments = pd.read_csv(RAW_PATH / "payments_raw.csv")
returns = pd.read_csv(RAW_PATH / "returns_raw.csv")
marketing = pd.read_csv(RAW_PATH / "marketing_campaigns_raw.csv")
```

```python
missing_cost = products[
    products["cost_price"].isna()
]

missing_cost
```

```python
products[["category", "cost_price", "list_price"]].groupby(
    "category"
).agg(
    average_cost=("cost_price", "mean"),
    average_list_price=("list_price", "mean"),
    products_with_cost=("cost_price", "count")
).round(2)
```

```python
products["cost_ratio"] = (
    products["cost_price"] / products["list_price"]
)
```

```python
category_cost_ratio = (
    products.groupby("category")["cost_ratio"]
    .median()
)

category_cost_ratio
```

```python
products_clean = products.copy()
```

```python
products_clean["cost_price_estimated"] = (
    products_clean["list_price"]
    * products_clean["category"].map(category_cost_ratio)
)
```

```python
products_clean["cost_price_final"] = products_clean["cost_price"]

missing_mask = products_clean["cost_price_final"].isna()

products_clean.loc[
    missing_mask,
    "cost_price_final"
] = products_clean.loc[
    missing_mask,
    "cost_price_estimated"
]
```

```python
products_clean["cost_price_source"] = np.where(
    products_clean["cost_price"].isna(),
    "Estimated",
    "Actual"
)
```

```python
print("Original missing costs:", products["cost_price"].isna().sum())

print(
    "Missing final costs:",
    products_clean["cost_price_final"].isna().sum()
)

print(
    "Estimated costs:",
    (products_clean["cost_price_source"] == "Estimated").sum()
)
```

```python
order_items_clean = order_items[
    order_items["quantity"] > 0
].copy()

print("Original rows:", len(order_items))
print("Clean rows:", len(order_items_clean))
print("Invalid rows removed:", len(order_items) - len(order_items_clean))
```

```python
missing_payment = payments[
    payments["payment_method"].isna()
].copy()

print("Missing payment methods:", len(missing_payment))

missing_payment
```

```python
order_payment_map = orders_clean.set_index(
    "order_id"
)["payment_method"]

payments_clean = payments.copy()

payments_clean["payment_method_recovered"] = (
    payments_clean["order_id"].map(order_payment_map)
)

payments_clean.loc[
    payments_clean["payment_method"].isna(),
    ["order_id", "payment_method", "payment_method_recovered"]
]
```

```python
recovery_mask = payments_clean["payment_method"].isna()

payments_clean.loc[
    recovery_mask,
    "payment_method"
] = payments_clean.loc[
    recovery_mask,
    "payment_method_recovered"
]

payments_clean["payment_method_source"] = np.where(
    recovery_mask,
    "Recovered from Orders",
    "Original"
)

print(
    "Recovered payment methods:",
    recovery_mask.sum()
)

print(
    "Remaining missing payment methods:",
    payments_clean["payment_method"].isna().sum()
)
```

```python
payments_clean = payments_clean.drop(
    columns=["payment_method_recovered"]
)

payments_clean.head()
```

```python
returns_clean = returns.copy()
marketing_clean = marketing.copy()

print("Returns rows:", len(returns_clean))
print("Marketing rows:", len(marketing_clean))
```

```python
customers_clean["date_of_birth"] = pd.to_datetime(
    customers_clean["date_of_birth"]
)

customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"]
)

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"]
)

payments_clean["payment_date"] = pd.to_datetime(
    payments_clean["payment_date"]
)

returns_clean["return_date"] = pd.to_datetime(
    returns_clean["return_date"]
)

marketing_clean["campaign_date"] = pd.to_datetime(
    marketing_clean["campaign_date"]
)

print("Date conversion completed.")
```

```python
clean_datasets = {
    "customers": customers_clean,
    "products": products_clean,
    "orders": orders_clean,
    "order_items": order_items_clean,
    "payments": payments_clean,
    "returns": returns_clean,
    "marketing": marketing_clean
}

for name, df in clean_datasets.items():
    print(f"{name:15} : {len(df):,} rows")
```

```python
key_checks_clean = {
    "customers": ("customer_id", customers_clean),
    "products": ("product_id", products_clean),
    "orders": ("order_id", orders_clean),
    "order_items": ("order_item_id", order_items_clean),
    "payments": ("payment_id", payments_clean),
    "returns": ("return_id", returns_clean),
    "marketing": ("campaign_id", marketing_clean)
}

for name, (key, df) in key_checks_clean.items():
    duplicates = df[key].duplicated().sum()

    print(
        f"{name:15} | "
        f"{key:15} | "
        f"duplicate keys: {duplicates}"
    )
```

```python
for name, df in clean_datasets.items():

    missing = df.isna().sum().sum()

    print(
        f"{name:15} | "
        f"missing values: {missing}"
    )
```

```python
customers_clean.to_csv(
    CLEAN_PATH / "customers_clean.csv",
    index=False
)

products_clean.to_csv(
    CLEAN_PATH / "products_clean.csv",
    index=False
)

orders_clean.to_csv(
    CLEAN_PATH / "orders_clean.csv",
    index=False
)

order_items_clean.to_csv(
    CLEAN_PATH / "order_items_clean.csv",
    index=False
)

payments_clean.to_csv(
    CLEAN_PATH / "payments_clean.csv",
    index=False
)

returns_clean.to_csv(
    CLEAN_PATH / "returns_clean.csv",
    index=False
)

marketing_clean.to_csv(
    CLEAN_PATH / "marketing_campaigns_clean.csv",
    index=False
)

print("All cleaned datasets saved successfully.")
```
